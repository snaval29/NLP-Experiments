```python
import nltk
import numpy as np
from nltk.corpus import treebank
from sklearn.model_selection import train_test_split

# Download dataset (first time only)

nltk.download('treebank')
```

```python
tagged_sentences = treebank.tagged_sents()

print(tagged_sentences[0])
```

```python
train_data, test_data = train_test_split(tagged_sentences, test_size=0.2)
```

```python
unigram_tagger = nltk.UnigramTagger(train_data)
```

```python
sentence = "Natural language processing is interesting".split()

print(unigram_tagger.tag(sentence))
```

```python
def rule_based_tagger(word):

    if word.endswith("ing"):
        return "VBG"

    elif word.endswith("ly"):
        return "RB"

    elif word[0].isupper():
        return "NNP"

    else:
        return "NN"
```

```python
sentence = "Machine learning is rapidly growing".split()

tags = [(word, rule_based_tagger(word)) for word in sentence]

print(tags)
```

```python
accuracy = unigram_tagger.evaluate(test_data)

print("ML POS Tagger Accuracy:", accuracy)
```
